## 2. Feature Engineering Review

This notebook reviews feature datasets already produced by the project. It does not regenerate features or change any files.

The available datasets may differ as the pipeline evolves, so the notebook checks which files exist before inspecting them.

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd()
if not (BASE_DIR / "data").exists():
    BASE_DIR = BASE_DIR.parent

PROCESSED = BASE_DIR / "data" / "processed"
CANDIDATES = [
    "customer_features.csv",
    "customer_churn_features.csv",
    "behavior_features.csv",
    "advanced_customer_features.csv",
    "clv_regression_data.csv",
    "ml_ready_data.csv",
    "tableau_customer_data.csv",
]

available = [PROCESSED / name for name in CANDIDATES
             if (PROCESSED / name).exists()]

print("Project:", BASE_DIR)
display(pd.DataFrame({
    "file": [p.name for p in available],
    "size_MB": [round(p.stat().st_size / (1024 * 1024), 2) for p in available]
}))
```

```python
# Inspect each available dataset using a bounded sample.
feature_samples = {}

for path in available:
    sample = pd.read_csv(path, nrows=10_000, low_memory=False)
    feature_samples[path.name] = sample
    print(f"\n{path.name}: sample shape = {sample.shape}")
    display(sample.head(3))
    print("Missing values in sample:", int(sample.isna().sum().sum()))
    print("Duplicate rows in sample:", int(sample.duplicated().sum()))
```

```python
# Review columns and likely target fields.
for filename, sample in feature_samples.items():
    print(f"\n{filename}")
    display(pd.DataFrame({
        "column": sample.columns,
        "dtype": sample.dtypes.astype(str).values,
        "missing_in_sample": sample.isna().sum().values,
        "unique_in_sample": sample.nunique(dropna=False).values,
    }))
    targets = [c for c in sample.columns
               if c.lower() in {"churn", "clv", "customer_id", "target"}]
    if targets:
        print("Potential target/identifier columns:", targets)
        for col in targets:
            if sample[col].nunique(dropna=False) <= 20:
                display(sample[col].value_counts(dropna=False).to_frame("count"))
```

## Interpretation checklist

- Confirm the expected customer identifier and target columns.
- Check for unexpected missing values and duplicate customer rows.
- Ensure identifiers and target labels are not accidentally used as predictive features.
- Confirm that the feature columns match the saved model's expected inputs before deployment.

This notebook is a review aid; it does not certify that the full dataset is valid based on a sample alone.
