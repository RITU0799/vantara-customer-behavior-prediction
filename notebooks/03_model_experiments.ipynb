## 3. Model Experiments and Evaluation Review

This notebook displays existing model-result CSV files for comparison. It does **not** train or tune any model.

Use the saved evaluation outputs as the source of truth. Check metric names and test/validation split labels before making comparisons.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = Path.cwd()
if not (BASE_DIR / "data").exists():
    BASE_DIR = BASE_DIR.parent

RESULTS_DIR = BASE_DIR / "data" / "processed" / "model_results"
print("Results folder:", RESULTS_DIR)
print("Folder exists:", RESULTS_DIR.exists())

result_files = sorted(RESULTS_DIR.glob("*.csv")) if RESULTS_DIR.exists() else []
display(pd.DataFrame({
    "file": [p.name for p in result_files],
    "size_KB": [round(p.stat().st_size / 1024, 1) for p in result_files]
}))
```

```python
# Display saved results, limiting rows for readability.
results_tables = {}

for path in result_files:
    try:
        table = pd.read_csv(path, nrows=200, low_memory=False)
        results_tables[path.name] = table
        print(f"\n### {path.name} — {table.shape[0]} displayed rows")
        display(table.head(20))
    except Exception as exc:
        print(f"Could not read {path.name}: {exc}")
```

```python
# Find common evaluation metrics in the loaded result tables.
metric_terms = [
    "roc_auc", "auc", "accuracy", "precision", "recall", "f1",
    "r2", "mae", "rmse", "mse"
]

for filename, table in results_tables.items():
    metric_columns = [
        col for col in table.columns
        if any(term in col.lower().replace("-", "_") for term in metric_terms)
    ]
    if metric_columns:
        print(f"\nMetric columns in {filename}: {metric_columns}")
        display(table[metric_columns].head(20))
```

```python
# Optional ROC-AUC comparison when one table has recognizable model and AUC columns.
for filename, table in results_tables.items():
    model_cols = [c for c in table.columns
                  if c.lower() in {"model", "model_name", "algorithm", "classifier"}]
    auc_cols = [c for c in table.columns
                if "roc_auc" in c.lower() or c.lower() == "auc"]
    if model_cols and auc_cols:
        comparison = table[[model_cols[0], auc_cols[0]]].copy()
        comparison[auc_cols[0]] = pd.to_numeric(
            comparison[auc_cols[0]], errors="coerce"
        )
        comparison = comparison.dropna()
        if not comparison.empty:
            display(comparison.sort_values(auc_cols[0], ascending=False))
            comparison.plot(
                x=model_cols[0], y=auc_cols[0],
                kind="bar", legend=False, rot=45
            )
            plt.title(f"Saved ROC-AUC comparison — {filename}")
            plt.ylabel("ROC-AUC")
            plt.tight_layout()
            plt.show()
            break
else:
    print("No result table with recognizable model and AUC columns was found.")
    print("Review the displayed result tables and their actual column names.")
```
